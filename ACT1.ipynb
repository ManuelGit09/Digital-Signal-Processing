## Part 3: Basic Signal Operations (Version 2)

```python
import numpy as np
import matplotlib.pyplot as plt


def stem_plot(n, y, title, ylabel, xlabel="n (sample index)", figsize=(8, 4)):
    """Reusable stem plot with grid and zero axes."""
    plt.figure(figsize=figsize)
    plt.stem(n, y)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True)
    plt.axhline(0, color="black", lw=0.5)
    plt.axvline(0, color="black", lw=0.5)
    plt.tight_layout()
    plt.show()


def show(title, **arrays):
    """Print a titled block of named arrays."""
    print(f"\n{title}")
    for name, arr in arrays.items():
        print(f"\t{name} =", arr.tolist())


n_x = np.arange(0, 5)
x = np.array([2, -1, 3, 0, 4])

# 1 Time Shift: y1[n] = x[n - 2]  -> indices move right by 2, values unchanged
n_y1, y1 = n_x + 2, x.copy()

# 2 Time Reversal: y2[n] = x[-n]  -> indices negated, values unchanged
n_y2, y2 = -n_x, x.copy()

# 3 Energy: E = sum |x[n]|^2
E = np.sum(np.abs(x) ** 2)

print("Original Sequence: x[n] = {2, -1, 3, 0, 4}")
print("\tn_x =", n_x.tolist())
print("\tx =", x.tolist())
show("#1 Time Shift: y1[n] = x[n - 2]", n_y1=n_y1, y1=y1)
show("#2 Time Reversal: y2[n] = x[-n]", n_y2=n_y2, y2=y2)
print("\n#3. Energy: E = sum |x[n]|^2")
print("\tEnergy E =", E)
```

```python
stem_plot(n_x, x, "Original Sequence x[n]", "x[n]")
stem_plot(n_y1, y1, "Time Shifted Sequence y1[n] = x[n - 2]", "y1[n]")
stem_plot(n_y2, y2, "Time Reversed Sequence y2[n] = x[-n]", "y2[n]")
```

# Discrete-Time Signal Processing: Analysis of an Arbitrary Finite-Duration Sequence

## 1. Overview
This document analyzes an **arbitrary, finite-duration discrete-time sequence** represented via explicit array values in Python using `numpy` and `matplotlib.pyplot`.

```python
n = np.arange(-5, 6)
x = np.array([0, 5, 6, 7, 9, 0, 0, -5, -9, -7, -6])
print(n)
print(x)
```

```python
plt.stem(n, x)
plt.xlabel('sampling interval')
plt.ylabel('sample x(n)')
plt.grid(True)
plt.show()
```

```python
def basic_signal(n, xn, title):
    """Stem plot used by the impulse / step / ramp / wave examples."""
    plt.stem(n, xn)
    plt.title(title)
    plt.xlabel('Sampling Interval')
    plt.ylabel('Sample')
    plt.grid(True)
    plt.show()


n9 = np.arange(-4, 5)   # same index range -4..4 for the three examples below

# Impulse: 1 only at n = 0
basic_signal(n9, (n9 == 0).astype(int), 'Impulse response')
```

```python
# Step: 1 for n >= 0
basic_signal(n9, (n9 >= 0).astype(int), 'Step response')
```

```python
# Ramp: x[n] = n
basic_signal(n9, n9, 'Ramp response')
```

```python
n = np.arange(-6.2, 6.3, 0.1)   # #1 and #2: np.arange(-3.1, 3.2, 0.1); #3 and #4: np.arange(-6.2, 6.3, 0.1)
xn = np.cos(n)                  # #1 np.sin(n)  #2 np.cos(n)  #3 np.sin(n)  #4 np.cos(n)
basic_signal(n, xn, 'Response Wave')   # Sine wave / Cosine wave
```
